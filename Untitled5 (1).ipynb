```python
# Online Food Ordering System

menu = {
    1: {"name": "Burger", "price": 120},
    2: {"name": "Pizza", "price": 250},
    3: {"name": "Pasta", "price": 180},
    4: {"name": "Sandwich", "price": 100},
    5: {"name": "Cold Drink", "price": 60}
}

cart = []

def show_menu():
    print("\n--- MENU ---")
    for key, item in menu.items():
        print(f"{key}. {item['name']} - ₹{item['price']}")

def add_to_cart():
    try:
        item_id = int(input("Enter item number to add: "))
        if item_id in menu:
            quantity = int(input("Enter quantity: "))
            cart.append((menu[item_id], quantity))
            print(f"{menu[item_id]['name']} added to cart!")
        else:
            print("Invalid item number!")
    except ValueError:
        print("Please enter valid input!")

def view_cart():
    print("\n--- YOUR CART ---")
    total = 0
    if not cart:
        print("Cart is empty!")
        return
    for item, qty in cart:
        cost = item["price"] * qty
        total += cost
        print(f"{item['name']} x {qty} = ₹{cost}")
    print(f"Total: ₹{total}")

def checkout():
    print("\n--- CHECKOUT ---")
    total = 0
    for item, qty in cart:
        total += item["price"] * qty
    print(f"Final Bill: ₹{total}")
    print("Thank you for ordering!")
    exit()

# Main program loop
while True:
    print("\n1. Show Menu")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        show_menu()
    elif choice == '2':
        show_menu()
        add_to_cart()
    elif choice == '3':
        view_cart()
    elif choice == '4':
        checkout()
    elif choice == '5':
        print("Exiting... Thank you!")
        break
    else:
        print("Invalid choice! Try again.")
```

```python
import datetime

menu = {
    1: {"name": "Burger", "price": 120},
    2: {"name": "Pizza", "price": 250},
    3: {"name": "Pasta", "price": 180},
    4: {"name": "Sandwich", "price": 100},
    5: {"name": "Cold Drink", "price": 60}
}

cart = []

# User name
user_name = input("Enter your name: ")
print(f"Welcome {user_name}! 🍽️")

def show_menu():
    print("\n--- MENU ---")
    for key, item in menu.items():
        print(f"{key}. {item['name']} - ₹{item['price']}")

def add_to_cart():
    try:
        item_id = int(input("Enter item number to add: "))
        if item_id in menu:
            quantity = int(input("Enter quantity: "))
            cart.append((menu[item_id], quantity))
            print(f"{menu[item_id]['name']} added to cart!")
        else:
            print("Invalid item number!")
    except ValueError:
        print("Please enter valid input!")

def view_cart():
    print("\n--- YOUR CART ---")
    total = 0
    if not cart:
        print("Cart is empty!")
        return
    for i, (item, qty) in enumerate(cart, start=1):
        cost = item["price"] * qty
        total += cost
        print(f"{i}. {item['name']} x {qty} = ₹{cost}")
    print(f"Subtotal: ₹{total}")

def remove_item():
    view_cart()
    try:
        index = int(input("Enter item number to remove: ")) - 1
        if 0 <= index < len(cart):
            removed = cart.pop(index)
            print(f"{removed[0]['name']} removed!")
        else:
            print("Invalid selection!")
    except ValueError:
        print("Enter valid number!")

def update_quantity():
    view_cart()
    try:
        index = int(input("Enter item number to update: ")) - 1
        if 0 <= index < len(cart):
            qty = int(input("Enter new quantity: "))
            item = cart[index][0]
            cart[index] = (item, qty)
            print("Quantity updated!")
        else:
            print("Invalid selection!")
    except ValueError:
        print("Enter valid input!")

def checkout():
    print("\n--- CHECKOUT ---")
    total = 0
    for item, qty in cart:
        total += item["price"] * qty

    gst = total * 0.05
    delivery = 40 if total < 300 else 0
    final_total = total + gst + delivery

    print(f"Subtotal: ₹{total}")
    print(f"GST (5%): ₹{gst:.2f}")
    print(f"Delivery Charge: ₹{delivery}")
    print(f"Final Bill: ₹{final_total:.2f}")

    print("\n--- ORDER SUMMARY ---")
    print(f"Customer: {user_name}")
    print("Time:", datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S"))

    print("\nThank you for ordering! 🎉")
    exit()

# Main loop
while True:
    print("\n1. Show Menu")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Remove Item")
    print("5. Update Quantity")
    print("6. Checkout")
    print("7. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        show_menu()
    elif choice == '2':
        show_menu()
        add_to_cart()
    elif choice == '3':
        view_cart()
    elif choice == '4':
        remove_item()
    elif choice == '5':
        update_quantity()
    elif choice == '6':
        checkout()
    elif choice == '7':
        print("Exiting... Thank you!")
        break
    else:
        print("Invalid choice! Try again.")
```
